```python
import requests
from datetime import datetime, timedelta
import json
import types
```

```python
# Replace with your API key
API_KEY = "DEMO_KEY"
NASA_BASE_URL = "https://api.nasa.gov/neo/rest/v1/feed"
```

```python
def get_neo(start_date=None, end_date=None, date_offset:int = 0):

    today = datetime.today()

    if date_offset >= 8:
        return "Date offset must be less than 8 days."

    if start_date is None:

        start_date = today
        start_date_api_format = start_date.strftime("%Y-%m-%d")
    else:
        start_date_api_format = datetime.strptime(start_date, "%Y-%m-%d").strftime("%Y-%m-%d")

    if end_date is None:
        end_date = today + timedelta(days=date_offset)
        end_date_api_format = end_date.strftime("%Y-%m-%d")
    else:
        end_date_api_format = datetime.strptime(end_date, "%Y-%m-%d").strftime("%Y-%m-%d")

    if (end_date - start_date).days > 8:
        return "Date range must be less than 8 days."

    params = {"start_date": start_date_api_format, "end_date": end_date_api_format, "api_key": API_KEY}

    response = requests.get(NASA_BASE_URL, params=params)
    api_data = response.json()
    return api_data
```
